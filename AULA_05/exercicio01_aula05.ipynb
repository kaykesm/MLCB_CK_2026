```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nltk
import spacy

from nltk.corpus import stopwords

nltk.download("stopwords", quiet=True)

stop_words_pt = set(stopwords.words("portuguese"))

nlp = spacy.load("pt_core_news_sm")

print("Ambiente preparado com sucesso.")
```

```python
def limpar_e_lemmatizar(texto):
    """
    Realiza o pré-processamento de uma mensagem.
    """

    # Converter para minúsculas
    texto_limpo = texto.lower()

    # Remover pontuação, números e caracteres especiais
    texto_limpo = re.sub(
        r'[^a-záàâãéèêíïóôõöúçñ\s]',
        '',
        texto_limpo
    )

    # Processar o texto com spaCy
    doc = nlp(texto_limpo)

    # Lista para armazenar os tokens finais
    tokens_filtrados = []

    for token in doc:

        # Ignorar espaços
        if token.is_space:
            continue

        # Remover stop words
        if token.text in stop_words_pt:
            continue

        # Remover tokens muito curtos
        if len(token.text) <= 1:
            continue

        # Adicionar o lema
        tokens_filtrados.append(token.lemma_)

    # Retornar o texto normalizado
    return " ".join(tokens_filtrados)
```

```python
frase_teste = (
    "Gostaria de saber se vocês estão DEVOLVENDO "
    "os valores das mesas compradas!!!"
)

resultado = limpar_e_lemmatizar(frase_teste)

print("Frase original:")
print(frase_teste)

print("\nFrase processada:")
print(resultado)
```

```python
def analisar_preprocessamento(texto):

    # Tokens antes do processamento
    tokens_antes = texto.split()

    # Aplicar o pré-processamento
    texto_normalizado = limpar_e_lemmatizar(texto)

    # Tokens depois do processamento
    tokens_depois = texto_normalizado.split()

    print("Texto original:")
    print(texto)

    print("\nTexto normalizado:")
    print(texto_normalizado)

    print("\nQuantidade de caracteres:")
    print(len(texto))

    print("\nQuantidade de tokens antes:")
    print(len(tokens_antes))

    print("\nQuantidade de tokens depois:")
    print(len(tokens_depois))

    print("\nTokens finais:")
    print(tokens_depois)
```

```python
analisar_preprocessamento(
    "MEU sofá!!! chegou quebrado e quero DEVOLVER!!!"
)
```

```python
def analisar_preprocessamento(texto):

    # Texto em minúsculas
    texto_lower = texto.lower()

    # Tokens originais
    tokens_antes = texto.split()

    # Remover pontuação e números
    texto_limpo = re.sub(
        r'[^a-záàâãéèêíïóôõöúçñ\s]',
        '',
        texto_lower
    )

    # Processar com spaCy
    doc = nlp(texto_limpo)

    tokens_finais = []
    tokens_removidos = []

    for token in doc:

        if token.is_space:
            continue

        if token.text in stop_words_pt:
            tokens_removidos.append(token.text)
            continue

        if len(token.text) <= 1:
            tokens_removidos.append(token.text)
            continue

        tokens_finais.append(token.lemma_)

    texto_normalizado = " ".join(tokens_finais)

    print("Texto original:")
    print(texto)

    print("\nTexto normalizado:")
    print(texto_normalizado)

    print("\nQuantidade de caracteres:")
    print(len(texto))

    print("\nQuantidade de tokens antes:")
    print(len(tokens_antes))

    print("\nQuantidade de tokens depois:")
    print(len(tokens_finais))

    print("\nTokens removidos:")
    print(tokens_removidos)

    print("\nTokens finais:")
    print(tokens_finais)
```

```python
analisar_preprocessamento(
    "MEU sofá!!! chegou quebrado e quero DEVOLVER!!!"
)
```
